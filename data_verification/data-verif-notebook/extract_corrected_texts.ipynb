# Slim Down Regenerated JSON Files

For each file in `fixed_batches_20_GSI/regenerate_from_scratch/`, keep only:
- `id`
- `specie`
- `specie_en`
- `corrected_text` (extracted from `verification.corrected_text`)

Each file is overwritten in place.

```python
import json
from pathlib import Path
```

```python
SOURCE_DIR = Path("../fixed_batches_20_GSI_all_entities/needs_text_regeneration_review")

json_files = sorted(SOURCE_DIR.glob("*.json"))
print(f"Found {len(json_files)} JSON files")
```

```python
for path in json_files:
    with open(path) as f:
        data = json.load(f)

    slim = {
        "id": data.get("id"),
        "specie": data.get("specie"),
        "specie_en": data.get("specie_en"),
        "corrected_text": data.get("verification", {}).get("corrected_text"),
    }

    with open(path, "w", encoding="utf-8") as f:
        json.dump(slim, f, ensure_ascii=False, indent=2)

    print(f"Updated: {path.name}")

print("\nDone.")
```

```python
for path in json_files:
    with open(path) as f:
        data = json.load(f)

    if "corrected_text" in data:
        data["fixed_text"] = data.pop("corrected_text")

        with open(path, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)

        print(f"Renamed corrected_text -> fixed_text: {path.name}")

print("\nDone.")
```

```python
SPECIES_DIR = SOURCE_DIR.parent / "species_folders"
SPECIES_DIR.mkdir(exist_ok=True)

for path in json_files:
    with open(path) as f:
        data = json.load(f)

    specie_name = data.get("specie", path.stem)
    folder = SPECIES_DIR / specie_name
    folder.mkdir(exist_ok=True)

    dest = folder / path.name
    with open(dest, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"Created: {folder.name}/{path.name}")

print("\nDone.")
```

```python
import shutil

SPECIES_DATA_DIR = Path("../../data-generation/species_data")

for path in json_files:
    with open(path) as f:
        data = json.load(f)

    specie_name = data.get("specie", path.stem)
    folder = SPECIES_DIR / specie_name

    src = SPECIES_DATA_DIR / f"{specie_name}.json"
    if src.exists():
        shutil.copy(src, folder / src.name)
        print(f"Copied {src.name} -> {folder.name}/")
    else:
        print(f"NOT FOUND: {src.name}")

print("\nDone.")
```
