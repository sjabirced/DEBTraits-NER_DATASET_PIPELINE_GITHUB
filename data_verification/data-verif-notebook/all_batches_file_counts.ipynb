# All Batches — File Count Summary

Counts `.json` files in each subfolder across all 20 batches in `sorted_batches_20_reprod_annot`:
- `with_errors`
- `without_errors`
- `with_reprod_annot`
- `without_reprod_annot`

```python
from pathlib import Path
import pandas as pd

base_dir = Path("../sorted_batches_20_reprod_annot")

subfolders = [
    "with_errors",
    "without_errors",
    "with_reprod_annot",
    "without_reprod_annot",
]

total_subfolders = ["with_errors", "without_errors", "without_reprod_annot"]

rows = []
for batch_dir in sorted(base_dir.iterdir()):
    if not batch_dir.is_dir() or not batch_dir.name.startswith("batch_"):
        continue
    row = {"Batch": batch_dir.name}
    for sub in subfolders:
        path = batch_dir / sub
        row[sub] = len(list(path.glob("*.json"))) if path.exists() else 0
    row["Total"] = sum(row[sub] for sub in total_subfolders)
    rows.append(row)

df = pd.DataFrame(rows)

# Totals row
totals = df[subfolders + ["Total"]].sum().to_dict()
totals["Batch"] = "TOTAL"
df_display = pd.concat([df, pd.DataFrame([totals])], ignore_index=True)

df_display
```

```python
import matplotlib.pyplot as plt
import numpy as np

batches = df["Batch"].tolist()
x = np.arange(len(batches))
width = 0.2

colors = {
    "with_errors":       "#d9534f",
    "without_errors":    "#5cb85c",
    "with_reprod_annot": "#5bc0de",
    "without_reprod_annot": "#f0ad4e",
}

fig, ax = plt.subplots(figsize=(18, 6))

for i, (sub, color) in enumerate(colors.items()):
    ax.bar(x + i * width, df[sub], width, label=sub, color=color)

ax.set_title("File counts per subfolder across all batches", fontsize=14, pad=12)
ax.set_ylabel("Number of JSON files")
ax.set_xlabel("Batch")
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(batches, rotation=45, ha="right")
ax.legend(loc="upper right")
plt.tight_layout()
plt.show()
```

```python
# Grand totals pie chart
grand = df[subfolders].sum()

fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(
    grand,
    labels=grand.index,
    autopct="%1.1f%%",
    colors=["#d9534f", "#5cb85c", "#5bc0de", "#f0ad4e"],
    startangle=140,
)
ax.set_title(f"All batches combined — {int(grand.sum())} total files", fontsize=13)
plt.tight_layout()
plt.show()
```
