# Find regenerated species across batch folders

For every species folder in `fixed_batches_20_GSI/regenerate_from_scratch_GSI/`,
search all 20 batch folders (`batch_01` to `batch_20`) and report in which batch each species exists.

```python
from pathlib import Path

base_dir      = Path("../fixed_batches_20_GSI")
regen_dir     = base_dir / "regenerate_from_scratch_GSI"

species_list  = sorted([d.name for d in regen_dir.iterdir() if d.is_dir()])
batch_dirs    = sorted([d for d in base_dir.iterdir() if d.is_dir() and d.name.startswith("batch_")])

results = []

for species in species_list:
    found_in = []
    for batch in batch_dirs:
        if (batch / species).is_dir():
            found_in.append(batch.name)
    results.append({"species": species, "found_in_batches": found_in})
    batches_str = ", ".join(found_in) if found_in else "NOT FOUND"
    print(f"{species:40s} -> {batches_str}")

print(f"\nTotal species in regenerate_from_scratch_GSI: {len(species_list)}")
```

# Replace species folders in batches with regenerated versions

For each species found in a batch folder, **delete** the existing species folder inside that batch
and **copy** the version from `regenerate_from_scratch_GSI` in its place.

```python
import shutil
from pathlib import Path

base_dir  = Path("../fixed_batches_20_GSI")
regen_dir = base_dir / "regenerate_from_scratch_GSI"

species_list = sorted([d.name for d in regen_dir.iterdir() if d.is_dir()])
batch_dirs   = sorted([d for d in base_dir.iterdir() if d.is_dir() and d.name.startswith("batch_")])

replaced = []

for species in species_list:
    src = regen_dir / species
    for batch in batch_dirs:
        dest = batch / species
        if dest.is_dir():
            # Remove the old folder and replace with the regenerated one
            shutil.rmtree(dest)
            shutil.copytree(src, dest)
            replaced.append({"species": species, "batch": batch.name})
            print(f"[REPLACED] {batch.name}/{species}")

print(f"\nDone. {len(replaced)} species folder(s) replaced.")
```
