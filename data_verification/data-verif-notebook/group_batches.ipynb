# Group Batch JSON Files

Collects all JSON files from `prompts_batches_20/batch_*/` into a single `dataset_ds/` folder.  
Keeps only: `id`, `specie`, `prompt`, and `corrected_text` (extracted from `verification.corrected_text`).

```python
import json
from pathlib import Path
```

```python
batches_dir = Path("../prompts_batches_20")
output_dir = Path("../verified_dataset_v1")
output_dir.mkdir(exist_ok=True)

json_files = sorted(batches_dir.glob("batch_*/*.json"))
print(f"Found {len(json_files)} JSON files across {len(list(batches_dir.glob('batch_*')))} batches")
```

```python
copied = 0
skipped = 0

for src in json_files:
    dst = output_dir / src.name
    if dst.exists():
        print(f"WARNING: duplicate filename '{src.name}' (from {src.parent.name}) — skipping")
        skipped += 1
    else:
        data = json.loads(src.read_text(encoding="utf-8"))
        record = {
            "id": data.get("id"),
            "specie": data.get("specie"),
            "prompt": data.get("prompt"),
            "corrected_text": data.get("verification", {}).get("corrected_text"),
        }
        dst.write_text(json.dumps(record, ensure_ascii=False, indent=2), encoding="utf-8")
        copied += 1

print(f"\nDone: {copied} files copied, {skipped} skipped (duplicates)")
print(f"Output folder: {output_dir.resolve()}")
```

```python
result_files = list(output_dir.glob("*.json"))
print(f"Files in all_batches/: {len(result_files)}")
```
