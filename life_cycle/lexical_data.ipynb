# Copy lifecycle JSON files to generated_dataset/lexical_data

Copies all `.json` files from `life_cycle/lifecycle_json/` into `generated_dataset/lexical_data/`.

```python
import shutil
from pathlib import Path

src_dir = Path('../life_cycle/lifecycle_json')
dst_dir = Path('lexical_data')

dst_dir.mkdir(parents=True, exist_ok=True)

json_files = list(src_dir.glob('*.json'))
print(f'Found {len(json_files)} JSON files in {src_dir}')

for f in json_files:
    shutil.copy2(f, dst_dir / f.name)

print(f'Copied {len(json_files)} files to {dst_dir}')
```
