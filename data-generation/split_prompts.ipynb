```python
import shutil
from pathlib import Path

SOURCE_DIR  = Path("prompts_to_send")
OUTPUT_DIR  = Path("prompts_batches_20")
BATCH_SIZE  = 370

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

files = sorted(SOURCE_DIR.glob("*.json"))
total = len(files)
n_batches = (total + BATCH_SIZE - 1) // BATCH_SIZE  # ceiling division

print(f"Total files   : {total}")
print(f"Total batches : {n_batches}")
print(f"Output folder : {OUTPUT_DIR}")
print()

for batch_idx in range(n_batches):
    batch_files = files[batch_idx * BATCH_SIZE : (batch_idx + 1) * BATCH_SIZE]
    batch_dir   = OUTPUT_DIR / f"batch_{batch_idx + 1:02d}"
    batch_dir.mkdir(exist_ok=True)

    for f in batch_files:
        shutil.copy2(f, batch_dir / f.name)

    print(f"batch_{batch_idx + 1:02d} : {len(batch_files)} files")

print()
print("Done.")
```
